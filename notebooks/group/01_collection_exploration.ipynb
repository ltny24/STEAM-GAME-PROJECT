## 3.1. Thu thập dữ liệu

Nhóm sử dụng bộ dữ liệu Steam Games tháng 03/2025 để khảo sát giá, cấu hình trò chơi, quy mô sở hữu và đánh giá. Phần này giới thiệu nguồn dữ liệu, cách tác giả thu thập và làm sạch, giấy phép và phạm vi sử dụng.

Các trường này giúp so sánh cách các game được phát hành và đón nhận trên Steam. Chẳng hạn, nhà phát triển có thể xem game có nhiều ngôn ngữ hoặc nhiều DLC thường đi cùng mức độ tương tác và quy mô sở hữu như thế nào.

```python
def check_zero_values(df):
    numeric_cols = df.select_dtypes(include="number").columns

    result = pd.DataFrame({
        "Số lượng giá trị 0": df[numeric_cols].eq(0).sum(),
        "Tỷ lệ giá trị 0 (%)": (
            df[numeric_cols].eq(0).mean() * 100
        ).round(2)
    })

    return result[result["Số lượng giá trị 0"] > 0]

zero_value_summary = check_zero_values(df)
display(zero_value_summary)

# Kiểm tra cả 47 cột GỐC trước khi drop: không bỏ qua cột sẽ bị loại.
audit_rows = []
for column in df.columns:
    values = df[column]
    numeric = pd.api.types.is_numeric_dtype(values)
    text_like = pd.api.types.is_object_dtype(values) or pd.api.types.is_string_dtype(values)
    stripped = values.astype('string').str.strip() if text_like else None
    audit_rows.append({
        'Cột gốc': column,
        'Kiểu gốc': str(values.dtype),
        'N có dữ liệu': int(values.notna().sum()),
        'NaN (%)': values.isna().mean() * 100,
        'Số giá trị khác nhau': values.nunique(dropna=True),
        '0/False (%)': values.eq(0).mean() * 100 if numeric else float('nan'),
        'Số mã -1': int(values.eq(-1).sum()) if numeric else float('nan'),
        'Chuỗi trắng': int(stripped.eq('').sum()) if text_like else float('nan'),
        'Chuỗi [] hoặc {}': int(stripped.isin(['[]', '{}']).sum()) if text_like else float('nan'),
        'Min (số gốc)': values.min() if numeric else float('nan'),
        'Max (số gốc)': values.max() if numeric else float('nan'),
    })
original_column_audit = pd.DataFrame(audit_rows).set_index('Cột gốc')
assert original_column_audit.index.tolist() == df.columns.tolist()
print(f'BẢNG KIỂM TOÀN BỘ {len(original_column_audit)} CỘT GỐC, {len(df):,} DÒNG - TRƯỚC LOẠI/GỘP')
print('NaN, chuỗi trắng và chuỗi danh sách rỗng được đếm riêng; -1/0 chưa bị tự thay đổi.')
with pd.option_context('display.max_rows', None, 'display.max_columns', None):
    display(original_column_audit.round(2))

# Đối chiếu theo từng bản ghi trước khi loại hai cột Metacritic.
metacritic_score_zero = df["metacritic_score"].eq(0).fillna(False)
metacritic_url_missing = (
    df["metacritic_url"].astype("string").str.strip().fillna("").eq("")
)
metacritic_pair_counts = pd.crosstab(
    metacritic_score_zero, metacritic_url_missing
).reindex(index=[True, False], columns=[True, False], fill_value=0)
metacritic_pair_counts.index = pd.Index(["Bằng 0", "Khác 0 hoặc thiếu"], name="Điểm trong CSV")
metacritic_pair_counts.columns = ["URL thiếu", "URL có"]
```

```python
from ast import literal_eval

rows_before, cols_before = df.shape
df["appid"] = df["appid"].astype("string")
original_dates = df["release_date"]
parsed_dates = pd.to_datetime(original_dates, errors="coerce")
invalid_dates = (original_dates.notna() & parsed_dates.isna()).sum()
df["release_date"] = parsed_dates.dt.date
print(f'Ngày phát hành không chuyển được: {invalid_dates:,}')
if invalid_dates:
    print('Ví dụ ngày lỗi:', original_dates[original_dates.notna() & parsed_dates.isna()].head(3).tolist())

list_cols = [
    "supported_languages",
    "full_audio_languages",
    "packages",
    "developers",
    "publishers",
    "categories",
    "genres",
]

def to_list(value, column):
    if isinstance(value, list):
        return value

    if value is None or value is pd.NA or (isinstance(value, float) and pd.isna(value)):
        return pd.NA

    try:
        parsed = literal_eval(value)

        if isinstance(parsed, dict):
            return [parsed] if column == "packages" else list(parsed.keys())

        return parsed if isinstance(parsed, list) else pd.NA
    except (ValueError, SyntaxError, TypeError):
        return pd.NA

for col in list_cols:
    original_values = df[col]
    missing_before = df[col].isna().sum()
    df[col] = df[col].map(lambda value: to_list(value, col))
    invalid_count = df[col].isna().sum() - missing_before
    print(f'{col}: {invalid_count:,} giá trị không chuyển được; {missing_before:,} giá trị thiếu ban đầu')
    if invalid_count:
        print('Ví dụ giá trị lỗi:', original_values[original_values.notna() & df[col].isna()].head(3).tolist())

# tags giữ cả tên nhãn và số phiếu trong dict, kể cả dict rỗng {}.
def parse_tag_votes(value):
    try:
        parsed = literal_eval(value) if isinstance(value, str) else value
    except (ValueError, SyntaxError, TypeError):
        return pd.NA
    if isinstance(parsed, list) and len(parsed) == 0:
        return {}  # CSV dùng [] cho game không có tags; không tạo NaN mới.
    return parsed.copy() if isinstance(parsed, dict) else pd.NA

original_tags = df['tags'].copy()
df['tags'] = original_tags.map(parse_tag_votes)
invalid_tags = original_tags.notna() & df['tags'].isna()
print(f'tags: {invalid_tags.sum():,} giá trị không chuyển được sang dict; '
      f'{original_tags.isna().sum():,} giá trị thiếu ban đầu')
if invalid_tags.any():
    print('Ví dụ tags không đọc được:', original_tags.loc[invalid_tags].head(3).tolist())
# Bản đối chiếu riêng để kiểm tra các ô EDA phía sau không sửa số phiếu.
tag_votes_reference = df['tags'].map(lambda value: value.copy() if isinstance(value, dict) else pd.NA)
print('Số game có dict tags:', df['tags'].map(lambda value: isinstance(value, dict)).sum())
print('Số game có tags rỗng:', df['tags'].map(lambda value: isinstance(value, dict) and len(value) == 0).sum())

string_cols = [
    "name",
    "estimated_owners"
]

df[string_cols] = df[string_cols].astype("string")
print(f'Chuyển kiểu: {rows_before:,} → {len(df):,} dòng; {cols_before} → {df.shape[1]} cột')
```

#### 3.2.2.3. Phát hiện giá trị ngoại lai và so sánh IQR, Z-score

So sánh tỷ lệ bị đánh dấu bởi hàng rào $1.5 \times IQR$ và ngưỡng $|Z| > 3$ trên bảy biến. Bảng tra cứu cực trị ghi AppID, tên game và giá trị lớn nhất để kiểm tra từng trường hợp.


```python
outlier_rows = []
for column in available_outlier_columns:
    values = numeric_frame[column].dropna()
    if values.empty:
        continue
    lower_quartile, upper_quartile = values.quantile([0.25, 0.75])
    interquartile_range = upper_quartile - lower_quartile
    lower_bound = lower_quartile - 1.5 * interquartile_range
    upper_bound = upper_quartile + 1.5 * interquartile_range
    iqr_mask = (values < lower_bound) | (values > upper_bound)

    standard_deviation = values.std(ddof=0)
    if standard_deviation == 0:
        z_score_mask = pd.Series(False, index=values.index)
    else:
        z_scores = (values - values.mean()) / standard_deviation
        z_score_mask = z_scores.abs() > 3

    outlier_rows.append({
        "Thuộc tính": column,
        "Số ngoại lai (IQR)": int(iqr_mask.sum()),
        "Tỷ lệ ngoại lai (IQR, %)": iqr_mask.mean() * 100,
        "Số ngoại lai (Z-score)": int(z_score_mask.sum()),
        "Tỷ lệ ngoại lai (Z-score, %)": z_score_mask.mean() * 100,
    })

outlier_comparison = pd.DataFrame(outlier_rows).set_index("Thuộc tính")
print("So sánh số lượng và tỷ lệ ngoại lai theo IQR và Z-score")
display(outlier_comparison.round(2))

# Tra cứu bản ghi cực trị; chỉ tạo bảng phụ, không thay đổi df.
rows_before_extreme_lookup = len(df)
extreme_units = {
    'price': 'USD',
    'dlc_count': 'DLC',
    'achievements': 'thành tựu',
    'average_playtime_forever': 'phút',
    'median_playtime_forever': 'phút',
}
extreme_rows = []
for column, unit in extreme_units.items():
    values = numeric_frame[column].dropna()
    if values.empty:
        continue
    maximum = values.max()
    # Giữ cả trường hợp nhiều game cùng đạt giá trị lớn nhất.
    for index in values.index[values.eq(maximum)]:
        extreme_rows.append({
            'Thuộc tính': column,
            'appid': df.at[index, 'appid'],
            'Tên game': df.at[index, 'name'],
            'Giá trị lớn nhất': maximum,
            'Đơn vị': unit,
        })
extreme_examples = pd.DataFrame(extreme_rows)
print('BẢNG TRA CỨU CÁC BẢN GHI CÓ GIÁ TRỊ LỚN NHẤT')
with pd.option_context('display.max_rows', None, 'display.max_columns', None,
                       'display.float_format', '{:,.2f}'.format):
    display(extreme_examples)
print(f'Tra cứu cực trị: {rows_before_extreme_lookup:,} -> {len(df):,} game trong dữ liệu chính; bảo toàn dữ liệu gốc.')
assert len(df) == rows_before_extreme_lookup
```

### 3.2.3. Các cột phân loại

Phần này xem nhãn nào phổ biến, nhãn nào ít gặp và những tên nào có nhiều cách viết. Với các cột có nhiều nhãn trên một game, cần phân biệt số game với tổng lượt gán nhãn khi đọc tỷ lệ.


```python
categorical_cols = [
    "required_age",
    "supported_languages",
    "full_audio_languages",
    "developers",
    "publishers",
    "categories",
    "genres",
    "estimated_owners",
    "tags",
    "os",
]

list_columns = {
    "supported_languages", "full_audio_languages", "developers",
    "publishers", "categories", "genres", "os"
}

def categorical_values(column):
    if column == 'tags':
        # Chỉ trích tên nhãn trên Series tạm; df['tags'] vẫn giữ dict có số phiếu.
        labels = df[column].map(lambda value: list(value.keys()) if isinstance(value, dict) else pd.NA)
        return labels.explode()
    return df[column].explode() if column in list_columns else df[column]

report = []

for col in categorical_cols:
    values = categorical_values(col)
    counts = values.value_counts()

    report.append({
        "Column": col,
        "Number of distinct values": len(counts),
        "Most frequent value": counts.index[0] if not counts.empty else None
    })

report = pd.DataFrame(report)
display(report)
```

**Mức độ tập trung của nhãn**

Tỷ lệ lớn nhất là số lượt của nhãn phổ biến nhất chia cho tổng lượt nhãn hợp lệ trong cột. Với cột nhiều nhãn, đây là tỷ lệ trên *lượt nhãn*, không phải tỷ lệ số trò chơi. Ví dụ, `Windows` chiếm 74,87% các lượt hệ điều hành nhưng xuất hiện ở 89.588/89.618 trò chơi. Với `tags`, tỷ lệ được tính từ lượt game mang nhãn, không phải số phiếu của nhãn.

| Tên cột | Số giá trị khác nhau | Giá trị xuất hiện nhiều nhất | Số lượng lớn nhất | Tỷ lệ lớn nhất (%) |
| --- | --- | --- | --- | --- |
| `required_age` | 16 | `0` | 88589 | 98.85 |

```python
import re

check_cols = [
    "supported_languages",
    "full_audio_languages",
    "developers",
    "publishers",
    "categories",
    "genres",
    "tags",
    "os"
]

def normalize_label(value):
    return re.sub(r"[\W_]+", "", str(value).strip().lower())

inconsistent_report = []

for col in check_cols:
    values = categorical_values(col)

    values = values.dropna().astype(str).str.strip()

    groups = values.groupby(values.map(normalize_label))

    for normalized, group in groups:
        variants = sorted(group.unique())

        if len(variants) > 1:
            inconsistent_report.append({
                "Tên cột": col,
                "Nhãn chuẩn hóa": normalized,
                "Các biến thể": variants
            })

inconsistent_report = pd.DataFrame(inconsistent_report)

display(inconsistent_report)
```

```python
summary = inconsistent_report.groupby("Tên cột").agg(
    so_nhom=("Nhãn chuẩn hóa", "nunique"),
    so_bien_the=("Các biến thể", lambda x: sum(len(v) for v in x)),
    vi_du=("Các biến thể", lambda x: "; ".join(
        " / ".join(map(str, v)) for v in list(x)[:3]
    ))
).reset_index()

summary.columns = [
    "Tên cột",
    "Số nhóm không nhất quán",
    "Số biến thể",
    "Một số ví dụ"
]

print(f"Tổng số dòng chi tiết: {len(inconsistent_report)}")
display(summary)
```

| `playtime_skew_ratio` | Thời lượng chơi trung bình chia cho trung vị khi trung vị dương. |
| `recent_total_rating_gap` | Tỷ lệ đánh giá tích cực gần đây trừ tỷ lệ toàn kỳ, tính bằng điểm phần trăm và chỉ có giá trị khi có đủ hai tỷ lệ. |


```python
# Chỉ tạo biến mô tả; không lọc game hoặc thay giá trị thiếu bằng 0.
rows_before, cols_before = df.shape

def count_list_col(value):
    return len(value) if isinstance(value, (list, tuple, set)) else np.nan

def parse_owners(value):
    if not isinstance(value, str):
        return np.nan
    match = re.fullmatch(r'\s*([\d,]+)\s*-\s*([\d,]+)\s*', value)
    if match is None:
        return np.nan
    lower, upper = [int(part.replace(',', '')) for part in match.groups()]
    return (lower + upper) / 2 if 0 <= lower <= upper else np.nan

df['platform_count'] = df['os'].map(count_list_col)
df['lang_count'] = df['supported_languages'].map(count_list_col)
df['full_audio_count'] = df['full_audio_languages'].map(count_list_col)
df['estimated_owners_mid'] = df['estimated_owners'].map(parse_owners)
print('Khoảng ownership không đọc được:', df['estimated_owners_mid'].isna().sum())

df['playtime_mean_median_gap'] = df['average_playtime_forever'] - df['median_playtime_forever']
# Mẫu số bằng 0 không xác định được tỷ số; không cộng 1 vào đơn vị phút.
positive_median = df['median_playtime_forever'].where(df['median_playtime_forever'] > 0)
df['playtime_skew_ratio'] = df['average_playtime_forever'] / positive_median
# Chênh lệch hai tỷ lệ cùng snapshot, không đặt tên là suy giảm uy tín.
df['recent_total_rating_gap'] = df['pct_pos_recent'] - df['pct_pos_total']

derived_cols = ['platform_count', 'lang_count', 'full_audio_count', 'estimated_owners_mid',
                'playtime_mean_median_gap', 'playtime_skew_ratio', 'recent_total_rating_gap']
display(df[derived_cols].describe().round(2))
print(f'Tạo biến: {rows_before:,} -> {len(df):,} dòng; {cols_before} -> {df.shape[1]} cột')
assert len(df) == rows_before
```

#### Bước 2 — Ma trận tương quan Spearman trên toàn bộ biến số

Heatmap hiển thị hệ số tương quan hạng Spearman giữa **22 cột số**. Mỗi hệ số dùng các game có giá trị hợp lệ ở cả hai cột và cần ít nhất 30 quan sát. Bảng kèm theo báo số quan sát hợp lệ của từng cặp. Chỉ hiển thị tam giác dưới; phần trên chứa các hệ số đối xứng tương ứng.


```python
# Tự lấy toàn bộ cột số hiện có; appid là định danh chuỗi, không đưa vào tương quan.
numeric_cols = df.select_dtypes(include='number').columns.tolist()
df_corr = df[numeric_cols].replace([np.inf, -np.inf], np.nan)
valid = df_corr.notna().astype('int32')
pairwise_n = valid.T.dot(valid)
corr_matrix = df_corr.corr(method='spearman', min_periods=30)
print(f'HEATMAP TOÀN BỘ {len(numeric_cols)} CỘT SỐ / {df.shape[1]} CỘT; {len(df):,} GAME')
print('Các cột số:', numeric_cols)
print('Các cột phi số được khảo sát bằng thống kê nhóm/bảng chéo:',
      df.columns.difference(numeric_cols).tolist())
print('N hợp lệ của từng cặp (đường chéo là N từng cột):')
print(pairwise_n.to_string())
print('Cột không đủ biến thiên:', df_corr.columns[df_corr.nunique() < 2].tolist())
print('Các biến hiệu/tỷ số có quan hệ đại số với thành phần; không dùng riêng rho để loại cột.')

fig, ax = plt.subplots(figsize=(22, 19))
sns.heatmap(corr_matrix, mask=np.triu(np.ones_like(corr_matrix, dtype=bool)),
            annot=True, fmt='.2f', annot_kws={'size': 7}, cmap='RdBu_r',
            center=0, vmin=-1, vmax=1, linewidths=0.3, ax=ax,
            cbar_kws={'label': 'Spearman rho (không đơn vị)'})
ax.set_title(f'Tương quan Spearman - toàn bộ {len(numeric_cols)} cột số, N riêng từng cặp')
ax.tick_params(axis='x', rotation=55, labelsize=9)
ax.tick_params(axis='y', rotation=0, labelsize=9)
fig.text(0.5, 0.01, 'Chỉ hiển thị tam giác dưới; phần trên chứa các hệ số đối xứng tương ứng.', ha='center', fontsize=11)
fig.tight_layout(rect=[0, 0.03, 1, 1])
fig.savefig(FIGURES_DIR / 'eda_correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
plt.close(fig)

# Kiểm tra ảnh hưởng của các game có cùng giá trị 0, chỉ ở mức EDA.
pt_cols = ['average_playtime_forever', 'median_playtime_forever']
pt_mask = df[pt_cols].gt(0).all(axis=1)
print(f'Thời lượng chơi, cả hai > 0: {len(df):,} -> {pt_mask.sum():,} dòng; '
      f'không dùng {len(df) - pt_mask.sum():,} dòng cho phép đối chiếu này.')
print('Rho toàn bộ:', corr_matrix.loc[pt_cols[0], pt_cols[1]])
print('Rho trên cặp thời lượng dương:', df.loc[pt_mask, pt_cols].corr(method='spearman').iloc[0, 1])
```

**Nhận xét ma trận tương quan Spearman**

- Trong 6.621 game có cả hai tỷ lệ đánh giá, tỷ lệ tích cực toàn kỳ và gần đây có hệ số Spearman **0,78**.
- Số DLC và tổng số thành tựu có tương quan thuận **0,30**. Bảng dùng tổng thành tựu của game, vì vậy chưa xác định được thành tựu nào thuộc DLC.
- Tương quan giữa thời lượng trung bình và trung vị là **0,9998** trên toàn bộ dữ liệu và **0,9254** trên 8.010 game có cả hai thời lượng dương. Chênh lệch cho thấy kết quả nhạy với việc có tính nhóm cùng bằng `0` hay không.


#### Bước 3 — Phân tích bảng chéo và cấu trúc phân nhóm dữ liệu phi số

So sánh nhóm thời gian phát hành với quy mô DLC, nhóm sở hữu với độ phủ hệ điều hành, và nhãn thể loại với nhóm phát hành hoặc quy mô DLC.


##### 3a. Phân tích cohort phát hành theo thời gian và chiến lược DLC

Lập bảng số game và tỷ lệ các nhóm DLC trong từng cohort phát hành. Bảng tóm tắt kèm theo báo trung vị giá, trung vị số DLC và số game có dữ liệu đánh giá.


```python
# EDA CỘT NGÀY: nhóm phát hành của các game trong cùng một snapshot.
eda_view = df.copy()
dates = pd.to_datetime(eda_view['release_date'], errors='coerce')
eda_view['release_cohort'] = pd.cut(
    dates.dt.year, bins=[-np.inf, 2009, 2014, 2019, 2022, 2025, np.inf],
    labels=['<=2009', '2010-2014', '2015-2019', '2020-2022', '2023-2025', '>2025']
).cat.add_categories(['Không rõ']).fillna('Không rõ')
eda_view['dlc_band'] = pd.cut(
    eda_view['dlc_count'], bins=[-0.5, 0, 1, 5, 20, np.inf],
    labels=['0', '1', '2-5', '6-20', '>20']
).cat.add_categories(['Không rõ']).fillna('Không rõ')
print('THỐNG KÊ NHÓM THEO NGÀY PHÁT HÀNH')
print(f'Tạo nhóm: {len(df):,} -> {len(eda_view):,} dòng; ngày không xác định: {dates.isna().sum():,}')
release_stats = eda_view.groupby('release_cohort', observed=True).agg(
    n_games=('appid', 'size'), n_price=('price', 'count'), median_price_usd=('price', 'median'),
    n_dlc=('dlc_count', 'count'), median_dlc=('dlc_count', 'median'),
    n_total_rating=('pct_pos_total', 'count'), n_recent_rating=('pct_pos_recent', 'count'))
display(release_stats.round(2))
release_dlc_counts = pd.crosstab(eda_view['release_cohort'], eda_view['dlc_band'])
release_dlc_pct = release_dlc_counts.div(release_dlc_counts.sum(axis=1), axis=0) * 100
display(release_dlc_counts)
display(release_dlc_pct.round(2))
assert release_dlc_counts.to_numpy().sum() == len(df)
print('Cohort là các game khác nhau theo ngày phát hành, không phải biến động theo thời gian của cùng game.')
ax = release_dlc_pct.plot.bar(stacked=True, figsize=(12, 6), colormap='viridis')
ax.set(title='Cơ cấu số DLC theo nhóm ngày phát hành tại snapshot tháng 3/2025',
       xlabel='Nhóm năm phát hành', ylabel='Tỷ lệ game trong nhóm (%)')
ax.set_xticklabels([f'{label}\nN={release_dlc_counts.loc[label].sum():,}'
                    for label in release_dlc_pct.index], rotation=0)
ax.legend(title='Số DLC', bbox_to_anchor=(1.02, 1), loc='upper left')
fig = ax.get_figure()
fig.tight_layout()
fig.savefig(FIGURES_DIR / 'eda_release_cohort_dlc.png', dpi=150, bbox_inches='tight')
plt.show()
plt.close(fig)
```

**Nhận xét cohort phát hành**

- Cohort 2023–2025 có 34.590 game, chiếm 38,6% snapshot. Các cohort có độ dài và thời gian tích lũy DLC khác nhau.
- Tại thời điểm chụp dữ liệu, 87,71% game phát hành giai đoạn 2023–2025 chưa có DLC, so với 74,08% ở giai đoạn 2010–2014.


##### 3b. Phân tích phân khúc người sở hữu và độ phủ nền tảng

Giữ nhãn khoảng của `estimated_owners`. Bảng chéo báo số game và tỷ lệ theo số hệ điều hành trong mỗi khoảng sở hữu; bảng tóm tắt kèm trung vị giá.


```python
# EDA CỘT CHUỖI: giữ nhãn khoảng ownership, không coi midpoint là số đo chính xác.
string_view = df.copy()
string_view['owners_interval'] = string_view['estimated_owners'].fillna('Không rõ')
string_view['platform_group'] = string_view['platform_count'].astype('Int64').astype('string').fillna('Không rõ')
owner_order = sorted(string_view['owners_interval'].unique(),
                     key=lambda label: (pd.isna(parse_owners(label)),
                                        parse_owners(label) if pd.notna(parse_owners(label)) else 0, label))
owners_stats = string_view.groupby('owners_interval', observed=True).agg(
    n_games=('appid', 'size'), median_price_usd=('price', 'median'),
    median_languages=('lang_count', 'median'), median_peak_ccu=('peak_ccu', 'median'),
    n_total_rating=('pct_pos_total', 'count'), n_recent_rating=('pct_pos_recent', 'count')
).reindex(owner_order)
owners_platform_counts = pd.crosstab(string_view['owners_interval'], string_view['platform_group']).reindex(owner_order)
owners_platform_pct = owners_platform_counts.div(owners_platform_counts.sum(axis=1), axis=0) * 100
print(f'THỐNG KÊ CHUỖI OWNERSHIP: {len(df):,} -> {len(string_view):,} dòng; không lọc game')
display(owners_stats.round(2))
display(owners_platform_counts)
display(owners_platform_pct.round(2))
assert owners_platform_counts.to_numpy().sum() == len(df)
print('Khoảng 0 - 0 và 0 - 20000 được giữ thành nhãn riêng; chưa suy luận nguyên nhân của giá trị 0.')
print('appid/name là định danh, không tạo nhóm tên game tùy ý để suy luận mối quan hệ.')
```

**Nhận xét phân khúc người sở hữu**

Khoảng `0 - 20000` có 59.320 game (66,19%), trong đó 79,42% hỗ trợ một hệ điều hành. Có thêm 8.418 game thuộc khoảng `0 - 0`, được giữ thành nhóm riêng. Một số khoảng sở hữu cao chỉ có một game nên tỷ lệ ở các khoảng này nhạy với từng quan sát.


##### 3c. Phân tích đặc trưng đa nhãn: Thể loại, Hệ điều hành và Gói nội dung

Đếm game theo từng nhãn và lập bảng chéo thể loại theo cohort phát hành và nhóm DLC. Với `packages`, đếm số gói trong danh sách của mỗi game.


```python
# EDA CỘT DANH SÁCH: mỗi game được tính tối đa một lần trong mỗi nhãn.
def label_values(value):
    if isinstance(value, dict):
        value = list(value.keys())  # Biến cục bộ; không thay dict/số phiếu trong df.
    if not isinstance(value, (list, tuple, set)):
        return ['<Thiếu/lỗi parse>']
    labels = sorted({str(item).strip() for item in value if str(item).strip()})
    return labels or ['<Danh sách rỗng>']

label_tables = {}
membership_frames = {}
for column in ['os', 'supported_languages', 'full_audio_languages', 'genres',
               'categories', 'tags', 'developers', 'publishers']:
    members = eda_view[['appid', column, 'price', 'estimated_owners_mid',
                  'pct_pos_total', 'pct_pos_recent', 'release_cohort', 'dlc_band']].copy()
    members[column] = members[column].map(label_values)
    members = members.explode(column).rename(columns={column: 'label'})
    assert members['appid'].nunique() == df['appid'].nunique()
    assert not members.duplicated(['appid', 'label']).any()
    table = members.groupby('label', observed=True).agg(
        n_games=('appid', 'size'), median_price_usd=('price', 'median'),
        n_owners=('estimated_owners_mid', 'count'), median_owners_mid=('estimated_owners_mid', 'median'),
        n_total_rating=('pct_pos_total', 'count'), n_recent_rating=('pct_pos_recent', 'count')
    ).sort_values('n_games', ascending=False)
    label_tables[column] = table
    membership_frames[column] = members[['appid', 'label', 'release_cohort', 'dlc_band']]
    print(f'{column}: {len(df):,} game -> {len(members):,} lượt game-nhãn '
          f'(+{len(members) - len(df):,} do nhiều nhãn); {len(table):,} nhãn.')
    print('Hiển thị tối đa 10 nhãn phổ biến; toàn bộ nhãn vẫn lưu trong label_tables.')
    display(table.head(10).round(2))

# Bảng chéo thể loại với nhóm ngày; explode làm game đa thể loại xuất hiện ở nhiều hàng.
genre_members = membership_frames['genres']
genre_cohort_counts = pd.crosstab(genre_members['label'], genre_members['release_cohort'])
genre_dlc_counts = pd.crosstab(genre_members['label'], genre_members['dlc_band'])
print('THỂ LOẠI x COHORT: số game trong từng thể loại; không cộng các hàng để suy ra số game duy nhất.')
display(genre_cohort_counts.reindex(label_tables['genres'].index).head(10))
print('THỂ LOẠI x NHÓM DLC: tỷ lệ hàng, hiển thị 10 thể loại phổ biến.')
display((genre_dlc_counts.div(genre_dlc_counts.sum(axis=1), axis=0) * 100)
        .reindex(label_tables['genres'].index).head(10).round(2))

# packages chứa dict lồng nhau: khảo sát số gói, không ép dict thành nhãn thể loại.
package_view = df[['appid', 'packages', 'price']].copy()
package_view['n_packages'] = package_view['packages'].map(count_list_col)
package_stats = package_view.groupby('n_packages', dropna=False).agg(
    n_games=('appid', 'size'), median_price_usd=('price', 'median'))
print(f'packages: {len(df):,} -> {len(package_view):,} dòng; thống kê số gói trong danh sách')
display(package_stats.round(2))
```

**Nhận xét cấu trúc đa nhãn và danh mục**

- Indie (63.189 game), Casual (38.699) và Action (36.842) là ba nhãn thể loại phổ biến nhất. Một game có thể xuất hiện ở nhiều nhóm.
- Windows có mặt ở 89.588 game (99,97%). Hỗ trợ macOS và Linux có thể cùng xuất hiện ở một game.
- 75.161 game (83,87%) có danh sách `packages` gồm một phần tử.


#### Bước 4 — Khảo sát quan hệ giữa độ phủ hỗ trợ, nội dung DLC và thành tựu

So sánh trung vị ownership ước tính theo độ phủ nền tảng và ngôn ngữ. Với DLC, báo cỡ mẫu, trung vị và tứ phân vị của tổng thành tựu theo năm nhóm số lượng DLC.


```python
# THỐNG KÊ NHÓM EDA độc lập với các bộ lọc theo câu hỏi trước đây.
group_view = df.copy()
group_view['language_group'] = pd.cut(
    group_view['lang_count'], bins=[-0.5, 0, 1, 5, 10, 20, np.inf],
    labels=['0', '1', '2-5', '6-10', '11-20', '>20']
).cat.add_categories(['Không rõ']).fillna('Không rõ')
group_view['dlc_group'] = pd.cut(
    group_view['dlc_count'], bins=[-0.5, 0, 1, 5, 20, np.inf],
    labels=['0', '1', '2-5', '6-20', '>20']
).cat.add_categories(['Không rõ']).fillna('Không rõ')
print(f'Tạo nhóm: {len(df):,} -> {len(group_view):,} game; gồm cả nhóm 0 và thiếu')
platform_stats = group_view.groupby('platform_count', dropna=False).agg(
    n_games=('appid', 'size'), n_owners=('estimated_owners_mid', 'count'),
    median_owners_mid=('estimated_owners_mid', 'median'), median_price_usd=('price', 'median'))
lang_stats = group_view.groupby('language_group', observed=True).agg(
    n_games=('appid', 'size'), n_owners=('estimated_owners_mid', 'count'),
    median_owners_mid=('estimated_owners_mid', 'median'))
dlc_stats = group_view.groupby('dlc_group', observed=True).agg(
    n_games=('appid', 'size'), n_achievements=('achievements', 'count'),
    zero_achievements=('achievements', lambda values: values.eq(0).sum()),
    achievements_q25=('achievements', lambda values: values.quantile(.25)),
    achievements_median=('achievements', 'median'),
    achievements_q75=('achievements', lambda values: values.quantile(.75)))
print('Nền tảng và ownership trung điểm:'); display(platform_stats.round(2))
print('Ngôn ngữ và ownership trung điểm:'); display(lang_stats.round(2))
print('Số DLC và tổng achievements (không quy achievements cho từng DLC):'); display(dlc_stats.round(2))
assert dlc_stats['n_games'].sum() == len(df)
fig, ax = plt.subplots(figsize=(10, 5))
ax.bar(dlc_stats.index.astype(str), dlc_stats['achievements_median'], color='steelblue', label='Trung vị')
ax.vlines(np.arange(len(dlc_stats)), dlc_stats['achievements_q25'], dlc_stats['achievements_q75'],
          color='black', linewidth=2, label='Q25-Q75 (không phải CI)')
ax.set(title='Tổng achievements theo nhóm số DLC - mô tả trên toàn bộ game',
       xlabel='Số DLC', ylabel='Achievements (số thành tựu/game)')
ax.set_xticks(np.arange(len(dlc_stats)), [f'{label}\nN={int(row.n_games):,}'
                                         for label, row in dlc_stats.iterrows()])
ax.legend()
fig.tight_layout()
fig.savefig(FIGURES_DIR / 'eda_dlc_achievements_groups.png', dpi=150, bbox_inches='tight')
plt.show()
plt.close(fig)
```

**Nhận xét thống kê nhóm**

- Các nhóm hỗ trợ một, hai và ba nền tảng đều có trung vị ownership trung điểm là 10.000. Các nhóm hỗ trợ ít nhất một ngôn ngữ cũng có trung vị này, nên cần xem thêm phân phối khi so sánh độ phủ.
- Trung vị tổng thành tựu lần lượt là **0, 17, 28, 40 và 42,5** cho các nhóm **0, 1, 2–5, 6–20 và trên 20 DLC**. Hai nhóm DLC cao nhất có 963 và 282 game. Đây là khác biệt quan sát giữa các nhóm game.
