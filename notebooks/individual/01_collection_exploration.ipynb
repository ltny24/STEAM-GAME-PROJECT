```python
from ast import literal_eval

df["appid"] = df["appid"].astype("string")
df["release_date"] = pd.to_datetime(
    df["release_date"], errors="coerce"
).dt.date

list_cols = [
    "supported_languages",
    "full_audio_languages",
    "packages",
    "developers",
    "publishers",
    "categories",
    "genres",
    "tags"
]

def to_list(value):
    if isinstance(value, list):
        return value

    if value is None or (isinstance(value, float) and pd.isna(value)):
        return []

    try:
        value = literal_eval(value)

        if isinstance(value, dict):
            return list(value.keys())

        return value if isinstance(value, list) else []
    except:
        return []

df[list_cols] = df[list_cols].apply(
    lambda col: col.map(to_list)
)

string_cols = [
    "name",
    "estimated_owners"
]

df[string_cols] = df[string_cols].astype("string")
```

`packages` không được đếm như một cột phân loại: mỗi giá trị là danh sách các từ điển (`dict`) chứa nhiều thuộc tính của gói sản phẩm. Muốn phân tích gói, trước hết cần chọn một thuộc tính cụ thể (chẳng hạn tên gói).


```python
categorical_cols = [
    "required_age",
    "supported_languages",
    "full_audio_languages",
    "developers",
    "publishers",
    "categories",
    "genres",
    "estimated_owners",
    "tags",
    "os",
]

list_columns = {
    "supported_languages", "full_audio_languages", "developers",
    "publishers", "categories", "genres", "tags", "os"
}

report = []

for col in categorical_cols:
    values = df[col].explode() if col in list_columns else df[col]
    counts = values.value_counts()

    report.append({
        "Column": col,
        "Number of distinct values": len(counts),
        "Most frequent value": counts.index[0] if not counts.empty else None
    })

report = pd.DataFrame(report)
display(report)
```

- **Mức độ cân bằng**:

Tỷ lệ lớn nhất là số lượt của nhãn phổ biến nhất chia cho tổng lượt nhãn hợp lệ trong cột. Với cột nhiều nhãn, đây là tỷ lệ trên *lượt nhãn*, không phải tỷ lệ số trò chơi. Ví dụ, `Windows` chiếm 74,87% các lượt hệ điều hành nhưng xuất hiện ở 89.588/89.618 trò chơi.

```python
import re

check_cols = [
    "supported_languages",
    "full_audio_languages",
    "developers",
    "publishers",
    "categories",
    "genres",
    "tags",
    "os"
]

def normalize_label(value):
    return re.sub(r"[\W_]+", "", str(value).strip().lower())

inconsistent_report = []

for col in check_cols:
    values = (
        df[col].explode()
        if df[col].apply(lambda x: isinstance(x, list)).any()
        else df[col]
    )

    values = values.dropna().astype(str).str.strip()

    groups = values.groupby(values.map(normalize_label))

    for normalized, group in groups:
        variants = sorted(group.unique())

        if len(variants) > 1:
            inconsistent_report.append({
                "Tên cột": col,
                "Nhãn chuẩn hóa": normalized,
                "Các biến thể": variants
            })

inconsistent_report = pd.DataFrame(inconsistent_report)

display(inconsistent_report)
```

```python
summary = inconsistent_report.groupby("Tên cột").agg(
    so_nhom=("Nhãn chuẩn hóa", "nunique"),
    so_bien_the=("Các biến thể", lambda x: sum(len(v) for v in x)),
    vi_du=("Các biến thể", lambda x: "; ".join(
        " / ".join(map(str, v)) for v in list(x)[:3]
    ))
).reset_index()

summary.columns = [
    "Tên cột",
    "Số nhóm không nhất quán",
    "Số biến thể",
    "Một số ví dụ"
]

print(f"Tổng số dòng chi tiết: {len(inconsistent_report)}")
display(summary)
```

```python
rare_cols = [
    "developers",
    "publishers",
    "categories",
    "genres",
    "tags",
    "os"
]

min_count = 180  # changable
rare_summary = []

for col in rare_cols:
    values = df[col].explode()
    counts = values.value_counts()

    rare = counts[counts < min_count]

    rare_summary.append({
        "Tên cột": col,
        "Ngưỡng min_count": min_count,
        "Số category ban đầu": len(counts),
        "Số category hiếm": len(rare),
        "Số category sau khi gom": len(counts) - len(rare) + (1 if not rare.empty else 0),
        "Số lượt bị gom": rare.sum(),
        "Tỷ lệ bị gom (%)": round(
            rare.sum() / counts.sum() * 100, 2
        )
    })

rare_summary = pd.DataFrame(rare_summary)

display(rare_summary)
```
