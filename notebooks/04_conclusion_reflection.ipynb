## 3.5.1. Tổng kết các phát hiện chính

Phần này tổng hợp và xâu chuỗi các phát hiện thực chứng từ sáu câu hỏi nghiên cứu cùng phát hiện bất ngờ nhất để phác họa bức tranh toàn cảnh về kinh tế học nền tảng, hành vi người tiêu dùng và chiến lược phát triển game trên Steam:

### 1. Câu hỏi 1 — Mở rộng độ phủ nền tảng, ngôn ngữ và quy mô người chơi (Phụ trách: Hồ Thị Như Ngọc)
* Khảo sát trên toàn bộ 89.618 tựa game cho thấy số lượng ngôn ngữ hỗ trợ có mối liên hệ đồng biến với lượng người sở hữu ước tính, nhưng ở mức độ yếu (Spearman $\rho = 0,153$; khoảng tin cậy bootstrap 95%: 0,146 – 0,161). Tương tự, số lượng hệ điều hành được hỗ trợ (Windows, macOS, Linux) cũng chỉ có liên hệ dương rất thấp với lượng người sở hữu ($\rho = 0,108$; khoảng tin cậy bootstrap 95%: 0,100 – 0,115). Kết quả này khẳng định việc chỉ đơn thuần bổ sung ngôn ngữ hay chuyển hệ sang đa nền tảng không đủ để bảo đảm sự tăng trưởng mạnh về quy mô người chơi.
* Mối liên hệ giữa số ngôn ngữ và lượng người sở hữu có sự phân hóa rõ rệt theo thể loại: thể hiện cao nhất ở Action ($\rho = 0,187$), Adventure ($\rho = 0,178$), RPG ($\rho = 0,158$), và thấp nhất ở Casual ($\rho = 0,036$). Khi phân tầng theo tuổi đời, tương quan giữa số ngôn ngữ và quy mô sở hữu tăng từ $0,119$ ở nhóm game mới (0 – 5 năm) lên $0,272$ ở nhóm trung bình (>5 – 10 năm) và đạt $0,351$ ở nhóm lâu năm (>10 năm). Điều này phản ánh cơ chế tích lũy vòng đời, các game duy trì được cộng đồng trung thành thường tiếp tục được nhà phát triển cập nhật thêm ngôn ngữ mới để khai thác thị trường quốc tế theo thời gian.
* Kết quả giữ nguyên chiều hướng tích cực khi thay đổi cách đại diện khoảng người sở hữu (mức thấp: $\rho = 0,180$; trung điểm: $\rho = 0,153$; mức cao: $\rho = 0,153$; loại bỏ khoảng `0–0`: $\rho = 0,195$; giới hạn giá trị cực lớn: $\rho = 0,153$). Quy trình tích hợp dữ liệu (D1) giữa file CSV chính và nguồn HuggingFace Parquet đạt tỷ lệ khớp 99,1% (88.813 / 89.618 game), với chỉ 0,5% (429 trường hợp) có sự khác biệt về số nền tảng, được ghi nhận minh bạch như một giới hạn của việc quan sát đa nguồn.

### 2. Câu hỏi 2 — Quy mô đánh giá và mức độ tham gia của người chơi (Phụ trách: Hồ Thị Như Ngọc)
* Trên tập mẫu 53.199 tựa game có dữ liệu đánh giá hợp lệ (sau khi loại 36.419 game mang mã thiếu `-1` do dưới 10 đánh giá), quy mô người sở hữu ước tính có mối tương quan thuận rõ rệt với tổng số lượt đánh giá (Spearman $\rho = 0,513$). Sau khi kiểm soát tuổi đời game và 6 flag thể loại chính bằng mô hình hồi quy, khi quy mô người sở hữu tăng gấp đôi, số lượt đánh giá tăng trung bình khoảng 11,5% (hệ số $\beta = 0,1567$; KTC 95%: 0,1506 – 0,1628), và mỗi năm tuổi đời game tích lũy thêm giúp tăng khoảng 4,3% lượt đánh giá ($\beta = 0,0425$; KTC 95%: 0,0375 – 0,0476).
* Khi số lượt đánh giá tăng gấp đôi, tỷ lệ đánh giá tích cực (`pct_pos_total`) chỉ tăng trong khoảng từ 0,9 đến 1,9 điểm phần trăm tùy theo thể loại: Simulation (+1,90 điểm %, KTC: 1,75 – 2,05), Indie (+1,79 điểm %, KTC: 1,67 – 1,90), Adventure (+1,57 điểm %, KTC: 1,44 – 1,70), RPG (+1,20 điểm %, KTC: 1,05 – 1,36), Casual (+1,18 điểm %, KTC: 1,04 – 1,32) và Action (+0,92 điểm %, KTC: 0,80 – 1,04). Kiểm định Wald cho thấy sự khác biệt về độ dốc giữa các thể loại có ý nghĩa thống kê cao ($p = 5,96 \times 10^{-53} < 0,001$). Nhưng, mức tăng thực tế là rất thấp (ví dụ từ 80% chỉ tăng lên lên 81–82%), bác bỏ quan niệm rằng game càng có nhiều người đánh giá thì sẽ càng có độ yêu thích cao.
* Khi kiểm tra độ bền bằng cách loại bỏ 1% game có số lượt đánh giá cao nhất thị trường (ngưỡng 33.601 lượt, còn lại 52.670 game), hệ số quy mô sở hữu vẫn duy trì ở mức 0,1333 (KTC 95%: 0,1279 – 0,1387), hệ số tuổi đời đạt 0,0458 (KTC 95%: 0,0410 – 0,0505), và mức tăng tỷ lệ tích cực vẫn ổn định trong biên độ 0,93 – 1,96 điểm phần trăm ($p < 0,001$). Điều này chứng minh quy luật quan sát được mang tính toàn hệ thống, không bị chi phối cục bộ bởi một vài hiện tượng đặc biệt.

### 5. Câu hỏi 5 — Mức độ tập trung lượng người sở hữu theo thể loại game (Phụ trách: Lê Thị Như Ý)
* **Cấu trúc thị trường kẻ thắng làm vua:** Trong thể loại Action, nhóm 10% tựa game dẫn đầu thâu tóm tới **90,1%** tổng lượng người sở hữu ước tính (và duy trì **84,2%** sau khi chuẩn hóa kiểm soát cùng thời kỳ phát hành và phân khúc giá). Các tựa game nhiều người chơi đầu bảng sở hữu hiệu ứng mạng lưới trực tiếp khổng lồ, tạo thành cấu trúc kim tự tháp nhọn hút trọn người chơi của phân khúc.
* **Tác động của chính sách Steam Direct từ năm 2017:** Ở các thể loại ngách (Indie, Casual), việc hạ phí niêm yết xuống 100 USD đã dẫn đến làn sóng dân chủ hóa xuất bản, làm loãng mức độ tập trung thị phần ở nhóm đuôi (tỷ trọng top 10% của Indie giảm từ 80,1% xuống 60,0%, Casual từ 76,1% xuống 47,2%). Tuy nhiên, pháo đài Action vẫn đứng vững ở mức xấp xỉ 86% qua mọi thế hệ nhờ rào cản gia nhập quá lớn.

### 6. Câu hỏi 6 — Nghịch lý thời lượng game: Lời hứa tiếp thị và thực tế trải nghiệm (Phụ trách: Lê Thị Như Ý)
* **Động lực định giá của nhà sản xuất:** Thời lượng cam kết dài đóng vai trò là đòn bẩy tiếp thị trực quan, giúp nhà phát hành biện minh cho mức giá bán cao hơn gấp 5 lần (từ 5,99 USD ở nhóm dưới 5 giờ lên 29,62 USD ở nhóm trên 50 giờ) và nâng quy mô sở hữu ban đầu lên bình quân 28,5% cho mỗi lần nhân đôi thời lượng.
* **Hố sâu tiêu thụ nội dung:** Game thủ vấp phải rào cản kiệt sức nhận thức khi đối mặt với game dài. Tỷ số tiêu thụ nội dung thực tế rơi thẳng đứng từ **1,16 ở nhóm dưới 5 giờ** xuống chỉ còn **0,24 ở nhóm trên 50 giờ**, đồng nghĩa hơn 83% người mua game rất dài không bao giờ trải nghiệm hết nửa sau của sản phẩm.
* **Hình phạt điểm đánh giá do gãy vỡ nhịp độ:** Kéo dài thời lượng một cách cơ học không làm tăng độ hài lòng mà gắn liền với mức giảm trung bình **-0,52 điểm phần trăm review tích cực** trên mỗi lần nhân đôi thời lượng cốt truyện sau khi kiểm soát các yếu tố gây nhiễu, đặc biệt giảm tới **-1,16 điểm phần trăm trong thể loại Action** do sự lặp lại của các thử thách cơ học gây ức chế tâm lý.

### 7. Phát hiện bất ngờ nhất trong toàn bộ nghiên cứu
* **Càng dài không đồng nghĩa với càng tốt:** Trái ngược với trực giác thông thường rằng game thủ luôn đòi hỏi thời lượng chơi tối đa cho số tiền bỏ ra, phân tích thực chứng chứng minh game thủ sẵn sàng phạt điểm nặng nề các tựa game cố tình kéo dài thời gian bằng các nhiệm vụ lặp lại thiếu chiều sâu. Sự tinh gọn và nhịp độ lôi cuốn trong khoảng **10 đến 20 giờ** đem lại tỷ lệ hài lòng cao hơn nhiều so với các sản phẩm kéo dài trên 50 giờ.
* Đa nền tảng không làm tăng lượng người dùng cho game mới và mâu thuẫn trong quy mô đánh giá:
  - Trái với niềm tin phổ biến rằng đổi game sang macOS và Linux sẽ mở rộng tập khách hàng, ở nhóm game mới phát hành (0 – 5 năm), liên hệ giữa số hệ điều hành và lượng người sở hữu gần như bằng 0 ($\rho = 0,020$). Windows vẫn là lựa chọn tuyệt đối của game PC.
  - Dù người sở hữu thúc đẩy số lượt đánh giá ($\rho = 0,513$), việc lượt đánh giá tăng gấp đôi chỉ kéo theo chưa tới 1–2 điểm phần trăm tỷ lệ tích cực. Thể loại đông người chơi nhất như Action lại có mức cải thiện điểm số thấp nhất (+0,92 điểm %), cho thấy khi cộng đồng mở rộng, kỳ vọng sẽ khiến điểm số dễ bị kìm hãm chứ không làm tăng điểm tích cực.

## 3.5.3. Giới hạn của đồ án

1. **Giới hạn về dữ liệu nguồn:**
   * **Bản chất ảnh chụp cắt ngang:** Dữ liệu là snapshot tại thời điểm tháng 03/2025, không phải dữ liệu chuỗi thời gian theo dõi cùng tựa game qua nhiều năm. Các so sánh thế hệ phản ánh sự khác biệt giữa các tựa game phát hành ở thời kỳ khác nhau chứ không đo lường trực tiếp quá trình suy giảm người chơi theo thời gian.
   * **Độ phân giải thô của biến người sở hữu:** Dữ liệu người sở hữu từ SteamSpy được lưu dưới dạng khoảng lũy thừa (ví dụ `0 - 20000`, `20000 - 50000`), làm mất đi độ biến thiên liên tục ở phân khúc game nhỏ.
   * **Thiên lệch tự báo cáo của HowLongToBeat:** Số liệu thời lượng phụ thuộc vào cộng đồng người chơi tự nguyện đóng góp (thường là những game thủ có kinh nghiệm và tâm huyết hơn bình thường), có thể ước lượng thời gian hoàn thành chuẩn hóa hơn so với người chơi đại chúng.
   * **Thiên lệch sống sót do cơ chế lọc ngưỡng đánh giá và dữ liệu thiếu:** Do cơ chế của Steam API ẩn thông tin các game có dưới 10 đánh giá, hai cột `pct_pos_total` và `num_reviews_total` có tới 40,64% bản ghi mang mã thiếu đặc biệt (-1). Việc loại bỏ 36.419 tựa game này trong mô hình Q2 tạo ra thiên lệch sống sót, khiến các ước lượng về mức độ tham gia đánh giá chỉ phản ánh phân khúc game đã vượt qua ngưỡng tối thiểu của cộng đồng. Bên cạnh đó, danh sách ngôn ngữ trống được ghi nhận là 0 ngôn ngữ không khẳng định được chắc chắn rằng trò chơi hoàn toàn không hỗ trợ ngôn ngữ nào.
2. **Giới hạn về phương pháp luận:**
   * **Mối liên hệ quan sát có điều kiện:** Các kết luận thống kê và hồi quy OLS phản ánh mối tương quan có điều kiện, chưa thể khẳng định quan hệ nhân quả tuyệt đối.
   * **Quy ước phân loại theo thể loại chính:** Việc gán mỗi tựa game vào thể loại đầu tiên giúp đảm bảo tính đơn nhất của đơn vị quan sát, nhưng đã đơn giản hóa đặc tính của các tựa game lai ghép đa thể loại.
