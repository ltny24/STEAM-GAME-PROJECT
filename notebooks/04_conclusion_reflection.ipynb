Các tựa game nhiều người chơi đầu bảng sở hữu hiệu ứng mạng lưới trực tiếp khổng lồ, tạo thành cấu trúc kim tự tháp nhọn hút trọn người chơi của phân khúc.
* **Tác động của chính sách Steam Direct từ năm 2017:** Ở các thể loại ngách (Indie, Casual), việc hạ phí niêm yết xuống 100 USD đã dẫn đến làn sóng dân chủ hóa xuất bản, làm loãng mức độ tập trung thị phần ở nhóm đuôi (tỷ trọng top 10% của Indie giảm từ 80,1% xuống 60,0%, Casual từ 76,1% xuống 47,2%). Tuy nhiên, pháo đài Action vẫn đứng vững ở mức xấp xỉ 86% qua mọi thế hệ nhờ rào cản gia nhập quá lớn.

### 6. Câu hỏi 6 — Nghịch lý thời lượng game: Lời hứa tiếp thị và thực tế trải nghiệm (Phụ trách: Lê Thị Như Ý)
* **Động lực định giá của nhà sản xuất:** Thời lượng cam kết dài đóng vai trò là đòn bẩy tiếp thị trực quan, giúp nhà phát hành biện minh cho mức giá bán cao hơn gấp 5 lần (từ 5,99 USD ở nhóm dưới 5 giờ lên 29,62 USD ở nhóm trên 50 giờ) và nâng quy mô sở hữu ban đầu lên bình quân 28,5% cho mỗi lần nhân đôi thời lượng.
* **Hố sâu tiêu thụ nội dung:** Game thủ vấp phải rào cản kiệt sức nhận thức khi đối mặt với game dài. Tỷ số tiêu thụ nội dung thực tế rơi thẳng đứng từ **1,16 ở nhóm dưới 5 giờ** xuống chỉ còn **0,24 ở nhóm trên 50 giờ**, đồng nghĩa hơn 83% người mua game rất dài không bao giờ trải nghiệm hết nửa sau của sản phẩm.
* **Hình phạt điểm đánh giá do gãy vỡ nhịp độ:** Kéo dài thời lượng một cách cơ học không làm tăng độ hài lòng mà gắn liền với mức giảm trung bình **-0,52 điểm phần trăm review tích cực** trên mỗi lần nhân đôi thời lượng cốt truyện sau khi kiểm soát các yếu tố gây nhiễu, đặc biệt giảm tới **-1,16 điểm phần trăm trong thể loại Action** do sự lặp lại của các thử thách cơ học gây ức chế tâm lý.

Sự tinh gọn và nhịp độ lôi cuốn trong khoảng **10 đến 20 giờ** đem lại tỷ lệ hài lòng cao hơn nhiều so với các sản phẩm kéo dài trên 50 giờ.

## 3.5.4. Nhìn lại quá trình thực hiện đồ án

### 1. Những khó khăn lớn nhất đã vượt qua
* **Xử lý cấu trúc dữ liệu bán cấu trúc:** Dữ liệu gốc chứa nhiều cột danh sách chuỗi (`genres`, `supported_languages`) và từ điển lồng nhau (`tags` kèm số phiếu bình chọn) bị lỗi định dạng; nhóm đã thiết lập đường ống phân tích cú pháp an toàn với `ast.literal_eval`.
* **Xử lý lạm phát số 0 và dữ liệu thiếu có điều kiện:** Hơn 91% game có thời lượng chơi bằng 0 và hơn 92% thiếu đánh giá gần đây. Nhóm đã không loại bỏ dữ liệu cảm tính mà phân tích đúng cơ chế vận hành của Steam API, áp dụng phân tầng mẫu có điều kiện và chuyển đổi logarit phù hợp.
* **Cộng tác Git nhóm và chuẩn hóa môi trường:** Trải qua các tình huống xung đột mã nguồn khi ghép nối các notebook; nhóm đã xây dựng quy trình phân nhánh Git, kiểm tra định dạng JSON nghiêm ngặt và quản trị môi trường Conda tái lập (`environment.yml`).

### 2. Những kiến thức và kỹ năng quý giá học được
* **Quy trình Khoa học Dữ liệu thực chiến:** Nắm vững toàn bộ vòng đời phân tích dữ liệu: từ thu thập, chuẩn hóa, khám phá đa biến, xây dựng mô hình hồi quy kinh tế lượng (OLS, sai số chuẩn vững HC3, kiểm định tương tác Wald, tái lấy mẫu Bootstrap), cho đến Machine Learning ngăn chặn triệt để rò rỉ dữ liệu.
* **Tư duy phản biện khoa học:** Không dừng lại ở việc chạy code và đọc lại số liệu đơn thuần, mà luôn đào sâu tìm kiếm cơ chế kinh tế và tâm lý học hành vi đằng sau, đồng thời kiểm chứng tính vững chắc qua nhiều kịch bản kiểm tra độ bền vững.
* **Kỹ năng làm việc nhóm chuyên nghiệp:** Phối hợp qua GitHub (Issues, Pull Requests, Code Review), thiết kế bài thuyết trình khoa học và tuân thủ liêm chính học thuật trong việc minh bạch sử dụng AI (`LLM_USAGE.md`).
